```python
ticket = {
    "id": 101,
    "customer": "Rangu",
    "is_urgent": True,
    "hours_open": 4.5
}

print(ticket)
print(ticket["customer"])
print(type(ticket))
```

```python
print(ticket["id"])
print(ticket["hours_open"])

ticket["status"] = "open"          # adds a NEW field
ticket["hours_open"] = 6.0         # overwrites an existing one

print(ticket)
```

```python
print(ticket["customer_name"])
```

```python
print(ticket.get("customer_name"))
print(ticket.get("customer_name", "unknown"))
```

```python
ticket2 = {
    "id": 102,
    "customer": "kohli",
    "is_urgent": False,
    "hours_open": 5.0,
    "status": "open",
    "priority": "High"
}
print(ticket2["status"])
print(ticket2["priority"])
print(ticket2.get("hours_open", "over"))
```

```python
priorities = ["High", "Medium", "Low"]

print(priorities)
print(priorities[0])
print(priorities[2])
print(len(priorities))
```

```python
print(priorities[3])
```

```python
priorities.append("Critical")   # adds to the END
print(priorities)

print(priorities[-1])           # LAST item, no counting needed
```

```python
tickets = [ticket, ticket2]

print(len(tickets))
print(tickets[0])
print(tickets[0]["customer"])
print(tickets[1]["priority"])
```
